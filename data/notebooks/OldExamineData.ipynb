### Novel method for determining  intractability during simulation

Below is the code I used to generate the coefficients for determining intractability in the `ndf_plus` data. I simply run a logistic regression on the `ndf` data (collected 26/05) and a linear regression over the predictions from it to estimate the benefits of a particular `quantile` selection. The logistic regression predicts whether a row will fall within the chosen `quantile` of the `seconds` column. If it does with probability `1 - quantile` or higher, I will declare it as intractable and terminate its computation. This will, on average, lead to the same total time spent, but now we are terminating earlier on runs we deemed as intractable in the old data. That extra time saved is used to explore new runs which were declared intractable before without taking much computation time. The result is a more rounded dataset with fewer time wasted exploring bad tree-algorithm pairings and more spent elsewhere.

### Limit to set of solved trees
To properly compare algorithms and see effects of different tree types on the different metrics we need to elminate a major source of bias. Simply dropping intractable runs will result in bias, because averages of different algorithms will be taken over different test instances with different independent variables. Therefore I will count how many trees solved different algorithms have in common in order to make some useful groupings.

Although the variable of *num-alts* seems a bit cryptic and implementation specific, which.. it is, its affects can easily be explained. The initial values for optimistic and pessimistic scores of nodes are randomly generated according to the maximum/minimum of $n$ uniformly distributed random variables. *Num-alts* is the $n$ in this equation. The higher the $n$, the wider the average gap between optimistic and pessimistic values will be. This means that children from all nodes will have wider gaps and thus take longer to converge to point-values. You can see it this way: the higher the $n$, the more certainty there is that the next state of the game will stay close to its previous state. Then it makes sense that high values for $n$ lead to less advantage for the player-to-move. The game goes on for longer, and there's less control to be gained from a single turn, so that first move has less of an impact on the final state of the game.

```python
xr = df_all.groupby(['range', 'growth']).sepValue.mean(); xr = xr / xr.index.get_level_values(0)
pd.DataFrame({'sepvals':xr}).reset_index().set_index('growth').sepvals.plot(marker='.', linestyle='none',label="mean - trees with same range")
xr.groupby('growth').mean().plot(label="mean - all trees");
plt.legend()
plt.xlabel("Tree growth factor")
plt.ylabel("Proven pessimistic value at root")
plt.title("Growth factor has inverse correlation with P2M advantage");
```

This graph is somewhat similar to the previous one: trees with a high growth factor will converge more slowly, thus leading to less of an advantage for the player-to-move. Important to note however is that the growth factor alone does not cause a tree to converge more quickly. This is mainly because a higher growth factor can also lead to a large number of irrelevant nodes being generated, which reduces the effective number of moves to choose from at internal nodes. Regardless, the pattern is strong enough to be clearly noticeable as seen above - but this probably does not tell the whole picture.

## Rapid-fire Boxplots

To keep computation times of these experiments reasonable there is a limit in place which terminates the search. This is at $30000$ nodes evaluated, $1000$ nodes of depth, or $4000$ nodes expanded.
