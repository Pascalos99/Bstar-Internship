## Finding incon seeds

```python
import pandas as pd
```

```python
df1 = pd.read_csv("../testing/quicktest.csv")
df2 = pd.read_csv("../testing/quicktest2.csv")
columns = [c for c in df1.columns if not c in ['seconds']]
gf1 = df1.groupby(['seed','algorithm'])[columns].first()
gf2 = df2.groupby(['seed','algorithm'])[columns].first()
dcolumns = gf1.columns[(gf1 != gf2).any()]
kf1 = gf1[(gf1 != gf2).any(axis=1)][dcolumns]
kf2 = gf2[(gf1 != gf2).any(axis=1)][dcolumns]
```

```python
rf1 = kf1.rename(columns={c : "%s1"%c for c in dcolumns})
rf2 = kf2.rename(columns={c : "%s2"%c for c in dcolumns})
comb = pd.concat([rf1, rf2], axis=1)[["%s%s"%(c,i) for c in dcolumns for i in '12']]
```

```python
comt = comb.copy()
for c in dcolumns:
    comt.loc[comb[comb[f'{c}1'] == comb[f'{c}2']].index, [f'{c}1',f'{c}2']] = ''
```

```python
incon_seeds = comt[comb['intractable1'] == "no"].reset_index().groupby('seed').algorithm.count().sort_values()[-1000:].index
```

## Comparing differences

```python
import pandas as pd
```

```python
df1 = pd.read_csv('../testing/incon_select_04.csv')
df2 = pd.read_csv('../testing/incon_select_05.csv')
df3 = pd.read_csv('../testing/incon_select_06.csv')
```

```python
columns = [c for c in df1.columns if not c in ['seconds']]
gf1 = df1.groupby(['seed','algorithm'])[columns].first()
gf2 = df2.groupby(['seed','algorithm'])[columns].first()
gf3 = df3.groupby(['seed','algorithm'])[columns].first()
dcolumns = gf1.columns[(gf1 != gf2).any() | (gf1 != gf3).any() | (gf2 != gf3).any()]
kf1 = gf1[((gf1 != gf2) | (gf1 != gf3) | (gf2 != gf3)).any(axis=1)][dcolumns]
kf2 = gf2[((gf1 != gf2) | (gf1 != gf3) | (gf2 != gf3)).any(axis=1)][dcolumns]
kf3 = gf3[((gf1 != gf2) | (gf1 != gf3) | (gf2 != gf3)).any(axis=1)][dcolumns]
rf1 = kf1.rename(columns={c : "%s1"%c for c in dcolumns})
rf2 = kf2.rename(columns={c : "%s2"%c for c in dcolumns})
rf3 = kf3.rename(columns={c : "%s3"%c for c in dcolumns})
comb = pd.concat([rf1, rf2, rf3], axis=1)[["%s%s"%(c,i) for c in dcolumns for i in '123']]
comt = comb.copy()
for c in dcolumns:
    comt.loc[comb[comb[f'{c}1'] == comb[f'{c}2']].index, [f'{c}1',f'{c}2']] = ''
    comt.loc[comb[comb[f'{c}1'] == comb[f'{c}3']].index, [f'{c}1',f'{c}3']] = ''
    comt.loc[comb[comb[f'{c}2'] == comb[f'{c}3']].index, [f'{c}2',f'{c}3']] = ''
```
