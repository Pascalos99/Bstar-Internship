```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("../testing/collected_data_supertest.csv"); df
```

```python
import csv

""" write the indices which are fully completed runs to a file so we can skip them in the next set of runs """

seed_counts = df.groupby(["seed"]).algorithm.count()

with open("completed_indices.csv","w") as f:
    wr = csv.writer(f,delimiter="\n")
    wr.writerow(list(seed_counts[seed_counts == 4].index))
```
