```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
```

### Loading the data

```python
main = pd.read_csv("../experiments/combined_data.zip")
main2 = pd.read_csv("../experiments/collected_data_alt_large.csv")
```

### Dropping new algorithms

```python
l1 = main.algorithm.unique()
l2 = main2.algorithm.unique()
extra_args = [x for x in l2 if not x in l1]
```

```python
main2F = main2[~main2.algorithm.replace({x : None for x in extra_args}).isna()]
```

### Sample maximum size

```python
run_count = main2F.groupby('seed').transform('count').min(axis=1)
```

```python
main3F = main2F[run_count == 16]
many_runs_gb = main3F.groupby(["style","growth","num_alts","range","width"])
sample_size = many_runs_gb.seed.transform(lambda x : len(x.unique())).min()
print("sample size is",sample_size)

def func(x):
    seeds = x.seed.sort_values().unique()
    return x.loc[x.seed.isin(seeds[:sample_size])].reset_index()

extra_rows = many_runs_gb.apply(func).set_index('index')
extra_rows.index.name = None

print()
test_gb = extra_rows.groupby(["style","growth","num_alts","range","width"])
display(test_gb.seed.transform(lambda x : len(x.unique())).describe())
print()
test_gb = extra_rows.groupby(["style","algorithm","growth","num_alts","range","width"])
display(test_gb.seed.transform(lambda x : len(x.unique())).describe())
print()
display(extra_rows.groupby('seed').transform('count').min(axis=1).describe())
```

### Join data and save to file

```python
super_join = pd.concat([main, extra_rows])
super_join = super_join.reset_index().drop(columns='index')
```

```python
name = f"combined_data_{100 + sample_size}"
super_join.to_csv(f"../experiments/{name}.zip", index=False, compression={'method':'zip', 'archive_name':f"{name}.csv"})
```
